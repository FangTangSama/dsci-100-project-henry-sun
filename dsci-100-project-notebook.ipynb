```r
# calculate mean, min, and max for each numeric variable
num_summary <- march_tidy |>
  summarise(
    mean_seed = mean(seed, na.rm = TRUE),
    min_seed = min(seed, na.rm = TRUE),
    max_seed = max(seed, na.rm = TRUE),
    mean_win_pct = mean(reg_wins_pct, na.rm = TRUE),
    min_win_pct = min(reg_wins_pct, na.rm = TRUE),
    max_win_pct = max(reg_wins_pct, na.rm = TRUE)
  )

num_summary
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).

```r
# count missing values for each column
missing_vals <- march_tidy |>
  summarise(
    year = sum(is.na(year)),
    school = sum(is.na(school)),
    seed = sum(is.na(seed)),
    reg_wins_pct = sum(is.na(reg_wins_pct)),
    bid = sum(is.na(bid)),
    advanced_first_round = sum(is.na(advanced_first_round))
  )

missing_vals
```

Visualization 1: This scatter plot displays the relationship between tournament seed and regular season win percentage, coloured by whether a team advanced past the first round. Teams that advanced (TRUE) are heavily concentrated in the top left area, showing lower seeds and higher win percentages. Teams eliminated in the first round (FALSE) mostly appear in the bottom right. This clear separation shows that these two predictors are relevant for predicting first round advancement with our model.

```r
ggplot(march_tidy, aes(x = seed, y = reg_wins_pct, colour = advanced_first_round)) +
  geom_point(alpha = 0.5) +
  labs(
    x = "Tournament Seed",
    y = "Regular Season Win Percentage",
    colour = "Advanced",
    title = "Seed vs Win Percentage by Outcome"
  )
```

Visualization 2: These faceted histograms compare the distribution of tournament seeds between teams that advanced (TRUE) and teams that were eliminated in the first round (FALSE). The advanced group is heavily skewed towards smaller seed numbers from 1 to 6, while the eliminated group is concentrated at higher seed numbers from 11 to 16. This demonstrates a strong relationship between a team's seed and their likelihood of winning in the first round.

```r
ggplot(march_tidy, aes(x = seed)) +
  geom_histogram(bins = 16) +
  facet_wrap(vars(advanced_first_round)) +
  labs(
    x = "Tournament Seed",
    y = "Count",
    title = "Distribution of Seeds by First Round Outcome"
  )
```
